```python
import pandas as pd
import numpy as np

# Définir les noms de colonnes pour le CSV initial à 17 colonnes (sans "Producteur")
columns_17 = [
    "Titre",
    "URL",
    "Entrées 1ère semaine",
    "Entrées totales",
    "Réalisateur",
    "Pays",
    "Genre",
    "Date de sortie",
    "Classification",
    "Distributeur",
    "Rang toutes exploitations",
    "Rang général",
    "Rang 2008",
    "Nb 1ère places hebdo",
    "Nb semaines Top20",
    "Meilleure place hebdo",
    "Acteurs"
]

# Charger le fichier invalid_17.csv sans en-tête et définir ses colonnes
df = pd.read_csv('invalid_17.csv', header=None, names=columns_17)

# Ajouter la colonne manquante "Producteur" et la remplir par NaN
df['Producteur'] = np.nan

# Normaliser les colonnes problématiques : remplissage des champs vides ou incohérents avec NaN
df = df.applymap(lambda x: np.nan if pd.isna(x) or (isinstance(x, str) and not x.strip()) else x)

# Réordonner les colonnes pour obtenir la structure souhaitée avec 18 colonnes
desired_columns = [
    "Titre",
    "URL",
    "Entrées 1ère semaine",
    "Entrées totales",
    "Réalisateur",
    "Producteur",  # Nouvelle colonne ajoutée
    "Pays",
    "Genre",
    "Date de sortie",
    "Classification",
    "Distributeur",
    "Rang toutes exploitations",
    "Rang général",
    "Rang 2008",
    "Nb 1ère places hebdo",
    "Nb semaines Top20",
    "Meilleure place hebdo",
    "Acteurs"
]

df = df[desired_columns]

# Sauvegarder le DataFrame modifié dans un fichier CSV final
df.to_csv('normalized_18.csv', index=False)

print("Les données ont été normalisées avec succès et sauvegardées dans 'normalized_18.csv'.")
```
