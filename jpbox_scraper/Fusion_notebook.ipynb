# 🔄 Film Data Merge & Backup Notebook

This notebook is designed to **safely merge** newly scraped film data (`films.csv`) with an existing backup dataset (`films_backup.csv`).

✅ **Purpose:**
- Prevent data loss after interruptions or crashes during long scraping sessions.
- Ensure that all unique films are preserved without duplication.

🔍 **What it does:**
1. Loads the existing backup and the newly scraped data.
2. Merges both datasets.
3. Identifies and reports how many duplicate `film_id`s exist between the two.
4. (Optional) Deduplicates the merged dataset based on `film_id`.
5. Saves the updated backup for future scraping continuity.

---

## 🔁 Merge Logic Overview

```mermaid
flowchart TD
    A[Start: Load films_backup.csv] --> B[Load films.csv]
    B --> C[Concatenate both datasets]
    C --> D[Check for duplicate film_id]
    D --> E{Keep duplicates?}
    E -- No --> F[Drop duplicates]
    E -- Yes --> G[Keep as-is]
    F --> H[Save updated films_backup.csv]
    G --> H
    H --> I[Use film_id list to avoid rescraping]


````python
import pandas as pd

# Charger les fichiers CSV
backup_df = pd.read_csv("films_backup.csv")
new_df = pd.read_csv("films.csv")

# Fusionner les deux avec un nouvel index
merged_df = pd.concat([backup_df, new_df], ignore_index=True)

# Repérer les doublons
duplicates = merged_df.duplicated(subset='film_id', keep=False)

# Afficher combien de doublons il y a
print(f"Nombre de lignes dupliquées basées sur 'film_id' : {duplicates.sum()}")

# Afficher les lignes dupliquées
merged_df[duplicates]
````

````python
# Combine and drop duplicates based on film_id
print(f"Nombre de lignes avant suppression des doublons : {len(merged_df)}")
print("Suppression des doublons...")
# Drop duplicates based on film_id, keeping the first occurrence
# and removing the rest
merged_unique = merged_df.drop_duplicates(subset="film_id", keep="first")
print(f"Nombre de lignes uniques après suppression des doublons : {len(merged_unique)}")

# Overwrite the backup with the updated data
merged_unique.to_csv("films_backup.csv", index=False, encoding="utf-8")
````
